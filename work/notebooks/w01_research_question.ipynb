## 3. Quick look at the data (2-3 real numbers)

Initial Data Insights:

Executing the code below inspects the anonymized metric distribution, calculating total dataset rows, baseline missingness, and the percentage of rows flagged as metric drop anomalies

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import os

data_path = 'data/raw/content_refresh_anonymized.csv'
if not os.path.exists(data_path):
    data_path = '../data/raw/content_refresh_anonymized.csv'

df = pd.read_csv(data_path)

total_pages = len(df)
print(f"1. Total Dataset Records: {total_pages:,}")

# Search for numeric delta columns to compute baseline anomaly rate
metric_col = next((c for c in ['clicks_change_pct', 'impressions_change_pct', 'delta_clicks'] if c in df.columns), None)

if metric_col:
    anomalies = df[df[metric_col] < 0]
    anomaly_pct = (len(anomalies) / total_pages) * 100
    print(f"2. Negative Delta Anomalies Detected: {len(anomalies):,} ({anomaly_pct:.1f}%)")
    print(f"3. Mean Anomaly Magnitude: {anomalies[metric_col].mean():.2%}")
else:
    print(f"2. Total Dataset Features: {df.shape[1]}")
    print(f"3. Total Missing Values: {df.isnull().sum().sum():,}")
```
